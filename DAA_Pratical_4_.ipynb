```python
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def factorial_recursive(n):

    if n <= 1:
        return 1


    return n * factorial_recursive(n - 1)


num = int(input("Enter a positive integer: "))

if num < 0:
    print("Factorial of a negative number is not defined.")
else:

    iterative_result = factorial_iterative(num)
    print("\n--- Iterative Method ---")
    print("Factorial of", num, "is:", iterative_result)


    recursive_result = factorial_recursive(num)
    print("\n--- Recursive Method ---")
    print("Factorial of", num, "is:", recursive_result)
```
